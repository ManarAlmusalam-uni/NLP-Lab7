# Part A — Decoder-Only Model

## Decoder-only example: GPT-style Text Generation

A decoder-only model predicts the next token based on previous tokens.

We will use **DistilGPT-2**, a small GPT-style model.


````python
generator = pipeline("text-generation", model="distilgpt2")

prompt = "Natural language processing is"

output = generator(prompt, max_new_tokens=30, do_sample=False)

print(output[0]["generated_text"])
````

# Part B — Encoder–Decoder Model

## Encoder–decoder example: Instruction / Text-to-Text Generation

Encoder–decoder models contain two main parts:

1. The **encoder** reads the input.
2. The **decoder** produces the output.

We will use **FLAN-T5-small**.


````python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small")

prompt = "Translate to French: I love natural language processing."

# Convert text into tokens
output = model.generate(**tokenizer(prompt, return_tensors="pt"))

# Convert tokens back to text
print(tokenizer.decode(output[0], skip_special_tokens=True))
````

# Part C — Encoder-Only Model

## Encoder-only example: Sentiment Classification

Encoder-only models read the full input and create a representation of it.

They are very useful when the task is to **understand or classify text**.

Here we use a DistilBERT sentiment model.


````python
classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")

text = "This NLP lab is easy and interesting."

result = classifier(text)
print(result)
````

## Task 3 — Encoder–Decoder Model

Use the FLAN-T5 model to perform the following:

- Summarize a short paragraph.


````python
paragraph = (
    "Transformers are a type of neural network architecture that rely on a mechanism called "
    "self-attention to process sequences of text. Unlike older recurrent models, they can look "
    "at all words in a sentence at the same time, which makes training faster and helps capture "
    "long-range dependencies. Transformers now power most modern language models, including "
    "BERT, T5 and GPT."
)

prompt = "Summarize: " + paragraph

inputs = tokenizer(prompt, return_tensors="pt", truncation=True)
summary_ids = model.generate(**inputs, max_new_tokens=60, num_beams=4, early_stopping=True)

print("Original paragraph:\n", paragraph)
print("\nSummary:\n", tokenizer.decode(summary_ids[0], skip_special_tokens=True))
````


## Task 5 — Short Reflection

Answer in 3–5 sentences:

**Why is a decoder-only model suitable for a chatbot such as ChatGPT?**


````python
# A decoder-only model is trained to predict the next token given everything that came before,
# which is exactly what a chatbot does: it reads the conversation so far and generates the reply one token at a time.
# Its causal (left-to-right) attention lets it generate text of any length autoregressively, feeding each new token back in as input.
# Because the user's message, the chat history and the reply can all be treated as one continuous sequence, no separate encoder is needed,
# which keeps the architecture simple and easy to scale to very large sizes and huge amounts of training data.
# Pre-training on next-token prediction followed by instruction tuning and human-feedback fine-tuning makes these models fluent,
#flexible and able to follow open-ended instructions.
````

### Question

Why do you sometimes get different outputs even when the prompt is the same?

**Hint:** Sampling introduces randomness when choosing the next token.


````python
# Run this cell several times: with sampling on and NO fixed seed, the output changes each run.
# (To get reproducible results again, uncomment the set_seed line.)
# set_seed(42)

prompt = "Natural language processing is"

for i in range(3):
    out = generator(prompt, max_new_tokens=30, do_sample=True, temperature=0.9, top_k=50,
                    pad_token_id=50256)
    print(f"Run {i+1}: {out[0]['generated_text']}\n")
````

**Answer:**

With `do_sample=True`, the model does not always pick the single most likely next token (as greedy decoding does). Instead it computes a probability distribution over the vocabulary and randomly samples from it (shaped by `temperature`, `top_k`, `top_p`). Each token drawn can differ between runs, and since every token affects all later predictions, the outputs diverge quickly. When `set_seed(42)` is set, the random number generator starts from the same state each time, so the same tokens get sampled and the output is reproducible. Greedy decoding (`do_sample=False`, as in Part A) is deterministic and always gives the same output for the same prompt.
